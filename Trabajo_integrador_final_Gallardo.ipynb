```python
#Entrenamiento del modelo.
modelo.fit(x_train,y_train)
```

```python
#Predicciones.
modelo.predict(x_test)
```

#**Evaluación del modelo**

---



```python
#R2 (Coeficiente de determinación)
from sklearn.metrics import r2_score
r2_score(y_test,modelo.predict(x_test))
```

```python
#MAE(Mean absolute error)
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y_test,modelo.predict(x_test))
```

```python
#RMSE (Root Mean Squared Error)
import numpy as np
rmse=np.sqrt(mean_absolute_error(y_test,modelo.predict(x_test)))
print (rmse)
```

*Nota: la MAE y RMSE indican que, si tenemos en cuenta un promedio de total sales del 3,5% aproximadamente, el promedio de equivocación del modelo es pequeño. Sin embargo, la exactitud es baja, probablemente porque las variables aportadas son pocas.*
